## 2. Charger les données

**Question marketing :** *de quoi disposons-nous pour répondre à notre question ?*

**La notion.** Un fichier de données est un **tableau** : une ligne par client, une colonne par information. Avant toute analyse, il faut le charger dans la mémoire du notebook. Nous utilisons pour cela un petit module fourni avec le cours, `preparation.py`, qui sait où se trouve le fichier et comment le lire. La cellule ci-dessous est **la première cellule de chaque notebook du cours** : vous la retrouverez à chaque séance.

```python
# Première cellule de chaque notebook : récupérer le module du cours, puis charger le fichier brut.
import os
import urllib.request

URL_MODULE = "https://raw.githubusercontent.com/2501Pr0ject/Data4Sup2Pub/main/notebooks/preparation.py"
if not os.path.exists("preparation.py"):                 # dans Colab le fichier n'est pas là : on le télécharge
    urllib.request.urlretrieve(URL_MODULE, "preparation.py")

import preparation as prep                                # le module est maintenant utilisable
brut = prep.charger()                                     # le tableau brut, tel que publié : rien n'est modifié
```

Le tableau s'appelle désormais `brut`. Ce nom est volontaire : c'est le fichier tel qu'il a été publié, sans aucune correction. Nous verrons en séance 3 qu'il contient des surprises.

## 3. Regarder le tableau

**Question marketing :** *combien de clients connaissons-nous, et que savons-nous de chacun ?*

**La notion.** Dans un tableau de données, **une ligne = une observation** (ici, un client) et **une colonne = une variable** (une information mesurée sur chaque client). Le nombre de lignes dit la taille de la base ; les noms de colonnes disent ce que l'enseigne a enregistré.

```python
# .shape donne (nombre de lignes, nombre de colonnes)
nb_lignes, nb_colonnes = brut.shape
print(f"{nb_lignes} lignes (une par client) et {nb_colonnes} colonnes (une information par colonne)")

brut.head()   # .head() affiche les 5 premières lignes
```

## 4. Une variable de près

**Question marketing :** *combien gagnent nos clients ?*

**La notion.** Pour une variable numérique, quelques chiffres suffisent à se faire une idée : le nombre de valeurs (`count`), la **moyenne** (`mean`), la **médiane** (`50%`, la valeur qui coupe la population en deux), le minimum et le maximum. Quand moyenne et médiane s'éloignent, c'est que quelques valeurs extrêmes tirent la moyenne.

```python
# .describe() résume une colonne numérique ; .round(0) arrondit pour la lisibilité
brut["Income"].describe().round(0)
```

**Question 3.** Comparez la moyenne et la médiane du revenu. Regardez ensuite le maximum, puis la ligne `count` : que remarquez-vous, et que faudrait-il vérifier avant de conclure quoi que ce soit ?

## 5. La variable que l'on voudra prédire

**Question marketing :** *qui a répondu à la dernière campagne ?*

**La notion.** Parmi toutes les colonnes, une a un statut à part : celle qu'on cherchera à **prédire**. On l'appelle la **cible**. Ici, `Response` vaut 1 si le client a accepté l'offre de la dernière campagne, 0 sinon. Sa moyenne est directement un **taux** (la part de 1).

```python
# value_counts() compte chaque valeur ; mean() sur des 0 et des 1 donne la part de 1
effectifs = brut["Response"].value_counts()
taux = brut["Response"].mean()
print(effectifs)
print(f"\nTaux de réponse : {taux:.1%}, soit {effectifs[1]} clients sur {len(brut)}")
```

## 6. Une variable en mots

**Question marketing :** *nos clients vivent-ils en couple ?*

**La notion.** Toutes les variables ne sont pas des nombres. `Marital_Status` est une variable **catégorielle** : elle prend un petit nombre de valeurs en texte, qu'on appelle des **modalités**. Pour la décrire, on compte les clients par modalité.

```python
# value_counts() : effectif de chaque modalité, de la plus fréquente à la plus rare
brut["Marital_Status"].value_counts()
```
